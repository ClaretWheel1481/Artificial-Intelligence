```python
!pip install jieba
import torch
import pandas as pd
import numpy as np
import jieba
import tensorflow as tf
import sys
from sklearn.model_selection import train_test_split
from tqdm import tqdm
```

```python
df = pd.read_csv('/content/drive/MyDrive/emotion_data/chnsenticorp/train.tsv',sep = '\t')
```

```python
# 获取数据的特征和标签
x = df['text_a']
y = df['label']
```

```python
df.groupby('label')['label'].count()
```

```python
text_cut = [jieba.lcut(text) for text in x]
```

```python
label_set = set()
for label in y:
  label_set.add(label)
label_set = np.array(list(label_set))

label_one_hot = []
for label in y:
  label_zero = np.zeros(len(label_set))
  label_zero[np.in1d(label_set,label)] = 1
  label_one_hot.append(label_zero)
labels = np.array(label_one_hot)
```

```python
tokenizer = tf.keras.preprocessing.text.Tokenizer()
tokenizer.fit_on_texts(text_cut)
```

```python
tokenizer.word_index
```

```python
text_seq = tokenizer.texts_to_sequences(texts=text_cut)
text_seq
```

```python
texts_pad_seq = tf.keras.preprocessing.sequence.pad_sequences(text_seq, maxlen=64, padding='post',truncating='post')
texts_pad_seq
```

```python
x_train, x_test, y_train, y_test = train_test_split(texts_pad_seq, labels, test_size=0.2,random_state=1)
```

```python
train_dataset = torch.utils.data.TensorDataset(torch.from_numpy(x_train),torch.from_numpy(y_train))
test_dataset = torch.utils.data.TensorDataset(torch.from_numpy(x_test), torch.from_numpy(y_test))
```

```python
batch_size = 32
train_loder = torch.utils.data.DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
test_loder = torch.utils.data.DataLoader(test_dataset, batch_size=batch_size, shuffle=True)
```

```python
class SentimentNet(torch.nn.Module):
  def __init__(self, layers, hidden_dim, embeding_dim, out_dim, vocab_size):
    super().__init__()
    self.layers = layers
    self.hidden_dim = hidden_dim
    self.embeding_dim = embeding_dim
    self.out_dim = out_dim
    self.lstm = torch.nn.LSTM(self.embeding_dim, self.hidden_dim, self.layers, dropout=0.5,
    batch_first=True)
    self.fc = torch.nn.Linear(in_features=self.hidden_dim, out_features=out_dim)
    self.sigmoid = torch.nn.Sigmoid()
    self.embeding = torch.nn.Embedding(vocab_size, self.embeding_dim)
    self.dropout = torch.nn.Dropout(0.5)

  def forward(self, x, hidden):
    x = torch.clamp(x, 0, self.embeding.num_embeddings - 1)

    embeds = self.embeding(x)
    lstm_out, hidden = self.lstm(embeds, hidden)
    out = self.dropout(lstm_out)
    out = self.fc(out)
    out = self.sigmoid(out)
    out = out[:,-1,:]
    out = out.squeeze()
    return out, hidden

  def init_hidden(self, batch_size):
    hidden = (torch.zeros(self.layers, batch_size, self.hidden_dim),
    torch.zeros(self.layers, batch_size, self.hidden_dim))
    return hidden
```

```python
model = SentimentNet(8, 128, 256, 2, 3000)
model.to('cpu')
```

```python
optimizer = torch.optim.Adam(model.parameters(), lr=0.0001)
# 二进制交叉熵损失函数
criterion = torch.nn.BCELoss()
```

# 模型训练

```python
epochs = 20
model.train()

for epoch in range(epochs):
  running_loss = 0
  running_acc = 0
  train_bar = tqdm(train_loder, file=sys.stdout)

  for step, (x_train, y_train) in enumerate(train_bar):
    cur_batch = len(x_train)
    # 初始化第一个hidden
    h = model.init_hidden(cur_batch)
    x_train = x_train.to('cpu')
    y_train = y_train.to('cpu')
    # 梯度清零
    model.zero_grad()
    output, h = model(x_train, h)
    #计算损失值
    loss = criterion(output, y_train.float())
    running_loss += loss.item()
    loss.backward()
    # 梯度剪裁
    torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=5)
    optimizer.step()
    # 统计准确率
    predict_y = torch.argmax(output, dim=1)
    running_acc += torch.eq(predict_y, y_train[:,1]).sum().item()
    train_bar.desc = f'train epoch[{epoch+1}/{epochs}]'
    train_bar.postfix = f'loss={running_loss} acc={running_acc}'
```
