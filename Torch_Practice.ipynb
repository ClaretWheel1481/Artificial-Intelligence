```python
import torch
```

创建一个全零的 5*5 张量

```python
torch.zeros(5,5)
```

创建一个从 1 到 10 的张量，并将其调整为2*5的矩阵

```python
torch.arange(1,11).reshape(2,5)
```

创建一个形状为(4,3)的单位矩阵

```python
torch.eye(4,3)
```

创建一个形状为(3,3)的常数张量，值为7

```python
torch.full((3,3),7)
```

使用 torch.ones_like() 创建与另一个张量相同形状的全 1 张量

```python
tensor = torch.tensor([[1,2],[3,4]])
torch.ones_like(tensor)
```

创建一个从 0 到 1 的 3*3随机张量

```python
torch.rand(3,3)
```

使用torch.full() 创建一个填充常数值的张量

```python
torch.full((2,2),5)
```

创建一个大小为(3,2)的张量，并将其填充成一个等差数列

```python
torch.arange(6).reshape(3,2)
```

使用torch.arange()创建一个包含10个数字的张量，步长为2

```python
torch.arange(0,20,2)
```
